Is notebook mein koi AI nahi chalta, koi API key nahi chahiye. Hum ginenge ki ek prompt AI ke liye **kitni baatein guess karne ko chhod deta hai**. Asli AI options gin ke nahi chalta, par idea wahi hai: jo tumne nahi bataya, wo AI guess karega.

```python
import itertools

OPTIONS = {
    "nazar":  ["hr expert", "career coach", "dost"],
    "kaun":   ["fresher", "5 saal", "manager"],
    "field":  ["data analyst", "teacher",
               "sales", "designer"],
    "skill":  ["python", "excel", "canva",
               "photoshop"],
    "length": ["1 page", "2 page"],
    "style":  ["bullet", "paragraph", "table"],
    "upar":   ["projects sabse upar",
               "education sabse upar",
               "experience sabse upar"],
}
sab = list(itertools.product(*OPTIONS.values()))
print(len(sab), "tarah ke resume ban sakte hain")
```

```python
def guesses(prompt):
    p, n, missing = prompt.lower(), 1, []
    for name, vals in OPTIONS.items():
        said = [v for v in vals if v in p]
        if not said:
            missing.append(name)
        n *= len(said) or len(vals)
    return n, missing

print(guesses("Resume bana do"))
```

```python
ROLE = "Tum 10 saal ke HR expert ho."
prompt = ROLE
print(guesses(prompt))
```

```python
CONTEXT = ("Main fresher hoon, Python aur SQL "
           "aata hai, data analyst job chahiye.")
prompt += " " + CONTEXT
print(guesses(prompt))
```

```python
TASK = ("Mere projects ke basis par 1 page ka "
        "resume banao, projects sabse upar.")
prompt += " " + TASK
print(guesses(prompt))
```

```python
FORMAT = ("Bullet points mein, har point "
          "10 words se kam.")
prompt += " " + FORMAT
print(guesses(prompt))
```

```python
def few_shot(examples, new):
    lines = ["Is style mein likho:"]
    for k, v in examples:
        lines.append(f"{k} -> {v}")
    lines.append(f"{new} ->")
    return "\n".join(lines)

ex = [("Diwali", "Roshni ka tyohar, 20% off!"),
      ("Holi", "Rangon ka tyohar, 20% off!")]
print(few_shot(ex, "Eid"))
```
