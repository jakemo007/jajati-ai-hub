Is notebook mein koi AI nahi chalta, koi API key nahi chahiye. Hum Python mein ek **chhota mock interviewer** banayenge: ek-ek sawal, aur har jawab par feedback. Feedback yahan simple rules se hai. Asli AI isse kahin behtar feedback deta hai, par loop wahi hai.

```python
SAWAL = [
    "Apne baare mein batao.",
    "Koi project batao jis par garv ho.",
    "Kisi mushkil deadline ka kissa sunao.",
    "Tumhari ek kamzori kya hai?",
    "Data analyst kyun banna hai?",
]

def list_do():          # list wala tareeka
    for q in SAWAL:
        print("-", q)

list_do()
print(len(SAWAL), "sawal ek saath. Ab ratta?")
```

```python
ROLE = ("Tum senior interviewer ho. Mera "
        "data analyst ka interview hai.")
EK_EK = ("Ek-ek karke sawal poochho, mere "
         "jawab ka wait karo.")
FEEDBACK = ("Har jawab ke baad feedback do: "
            "kya achha tha, kya sudharna hai.")

prompt = "\n".join([ROLE, EK_EK, FEEDBACK])
print(prompt)
```

```python
def interviewer(sawal):
    for n, q in enumerate(sawal, 1):
        yield f"Q{n}: {q}"

ai = interviewer(SAWAL)
print(next(ai))      # sirf pehla sawal
```

```python
print(next(ai))      # jawab ke baad, agla
```

```python
import re

def feedback(jawab):
    j, w = jawab.lower(), len(jawab.split())
    achha, sudhaaro = [], []
    if re.search(r"\d", j):
        achha.append("number diya")
    else:
        sudhaaro.append("koi number/result do")
    if "maine" in j:
        achha.append("apna example diya")
    else:
        sudhaaro.append("apna example do")
    if 25 <= w <= 90:
        achha.append(f"{w} words, sahi length")
    else:
        sudhaaro.append(f"{w} words: 25-90 rakho")
    return achha, sudhaaro

def dikhao(jawab):
    a, s = feedback(jawab)
    print("ACHHA:   ", ", ".join(a) or "-")
    print("SUDHAARO:", ", ".join(s) or "-")
```

```python
j1 = "Main hardworking hoon aur team player hoon."
dikhao(j1)
```

```python
JAWAB = [   # Colab mein input() se khud likho
    j1,
    "Maine college fest ke 500 tickets ka data "
    "Excel mein track kiya. Har din ki sale ka "
    "chart banaya, isliye team ne last 2 din "
    "promotion badhaya aur saare tickets bik gaye.",
    "Ek baar deadline mushkil thi, par maine raat "
    "bhar kaam karke time par de diya.",
]

def mock(jawab):
    strong = 0
    for q, j in zip(interviewer(SAWAL), jawab):
        a, s = feedback(j)
        print(q, "->", len(a), "achha,",
              len(s), "sudhaaro")
        strong += not s
    print("STRONG JAWAB:", strong, "/", len(jawab))

mock(JAWAB)
```

```python
JAWAB[0] = ("Main B.Com fresher hoon, mujhe "
    "SQL aur Excel aata hai. Maine college "
    "project mein 2,000 rows ka sales data saaf "
    "karke ek dashboard banaya. Ab mera goal "
    "data analyst ki pehli job hai.")
JAWAB[2] = ("Final year project ki report 3 din "
    "mein deni thi. Maine kaam 3 hisson mein "
    "baanta, roz 1 hissa poora kiya, aur report "
    "time se 4 ghante pehle de di.")
mock(JAWAB)
```
