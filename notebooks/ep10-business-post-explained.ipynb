Is notebook mein koi AI nahi chalta, koi API key nahi chahiye. Hum Python se ek post ki **aawaz naapenge**, AI ke liye 3 hisson wala prompt banayenge, aur 3 options mein se ek chunenge. Asha Home Bakery ek nakli dukaan hai, aur saare posts haath se likhe examples hain.

```python
# AI wala post (haath se likha example)
ai_post = ("Unlock the magic of premium taste! "
           "Discover our exclusive range of cakes, "
           "crafted with love and passion. "
           "Elevate every celebration today! "
           "✨🎂🎉🔥💯")

# Asha Home Bakery (nakli dukaan) ki post
meri_post = ("Kal 12 chocolate cake bane, 11 bik "
             "gaye 😄 Bacha hua 1 maine khud kha "
             "liya. Kal phir banenge, 2 baje tak "
             "order kar do.")

POSTS = {"AI": ai_post, "meri": meri_post}
```

```python
import re

CORP = ["unlock", "magic", "premium", "discover",
        "exclusive", "elevate"]   # company shabd

def style(post):
    p = post.lower()
    return {
        "emoji": sum(ord(c) > 0x2600 for c in post),
        "!": post.count("!"),
        "number": len(re.findall(r"\d+", post)),
        "corporate": sum(w in p for w in CORP),
    }

print("      emoji  !  number  corporate")
for naam, post in POSTS.items():
    e, x, n, c = style(post).values()
    print(f"{naam:4}  {e:5}  {x}  {n:6}  {c:9}")
```

```python
BUSINESS = ("Asha Home Bakery: ghar ke bane "
    "cake, Pune ke parivaaron ke liye. "
    "Offer: Sunday ko 10% off.")
EXAMPLE = (f"Meri purani post: {meri_post} "
    "Isi style mein likho.")
FORMAT = ("3 options do, har caption 2 lines "
    "ka, ek saaf call to action ke saath.")

prompt = "\n".join([BUSINESS, EXAMPLE, FORMAT])
print(prompt)
```

```python
# Maan lo AI ne ye 3 options diye
# (ye bhi haath se likhe examples hain)
OPTIONS = [
    "Sunday special! Unlock 10% off on "
    "premium cakes ✨🎉",
    "Sunday ko har cake par 10% off 😄 Kal 12 "
    "bane the, is baar 20 banenge.\n"
    "Order 2 baje tak kar do.",
    "Cake chahiye? Sunday ko 10% off.\n"
    "Bas itna hi.",
]
for n, o in enumerate(OPTIONS, 1):
    print(f"option {n}:", o.replace("\n", " / "))
```

```python
def score(post):
    s, m = style(post), style(meri_post)
    return sum([
        s["corporate"] == 0,   # company shabd nahi
        s["number"] > 0,         # asli number
        abs(s["emoji"] - m["emoji"]) <= 1,
        "order" in post.lower(), # call to action
        len(post.splitlines()) == 2,   # 2 lines
    ])

for n, o in enumerate(OPTIONS, 1):
    print(f"option {n}: {score(o)} / 5")
best = max(OPTIONS, key=score)
print("sabse achha: option",
      OPTIONS.index(best) + 1)
```

### Asli AI ke saath
`prompt` aur apni ek purani post ChatGPT ya Gemini mein paste karo. 3 options aayenge, **chunna tumhara kaam hai**. Post karne se pehle **offer, price aur timing khud check karo**: AI kabhi-kabhi aisa offer likh deta hai jo tumne diya hi nahi.
