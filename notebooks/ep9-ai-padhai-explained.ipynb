```python
def compound(p, rate, saal):
    for s in range(1, saal + 1):
        p = p * (1 + rate / 100)
        print(f"saal {s}: Rs {round(p, 2):g}")
    return p

paisa = compound(100, 10, 2)
simple = 100 + 100 * 10 / 100 * 2
print(f"simple interest: Rs {simple:g}")
```

```python
print("saal  compound  simple  farak")
for s in range(1, 11):
    c = 100 * 1.1 ** s      # compound
    si = 100 + 10 * s       # simple
    bar = "#" * int((c - si) // 4)
    print(f"{s:4}  {c:8.2f}  {si:6}  {bar}")
```

```python
def tricks(prompt):
    p, kami = prompt.lower(), []
    if "example" not in p:
        kami.append("example")
    if "practice" not in p:
        kami.append("practice")
    return kami

print(tricks("Compound interest ka answer do"))
```

```python
def padhai_prompt(topic):
    return (f"Mujhe {topic} samjhao. "
            "Ise real-life example se samjhao. "
            "Phir 3 practice sawal do, "
            "answers sabse last mein.")

p = padhai_prompt("compound interest")
print(p)
print("kami:", tricks(p))
```

```python
SAWAL = [(200, 10, 2), (1000, 5, 2), (500, 20, 3)]
for n, (p, r, s) in enumerate(SAWAL, 1):
    print(f"Q{n}: Rs {p}, {r}% compound, {s} saal?")
print("Video pause karo. Pehle khud try karo!")
```

```python
# ANSWERS (sabse last mein)
for n, (p, r, s) in enumerate(SAWAL, 1):
    jawab = p * (1 + r / 100) ** s
    print(f"Q{n}: Rs {round(jawab, 2):g}")
```
