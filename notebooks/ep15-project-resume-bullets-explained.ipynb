Is notebook mein koi asli AI nahi chalta, koi API key nahi chahiye. Ek chhota **bullet checker** hai: bullet mein verb, number aur tareeka dekhta hai, aur **aisa number pakadta hai jo tumhare facts mein nahi hai**. Project, bullets aur checker teaching ke liye bane hain, asli resume tool ke nahi.

```python
FACTS = {"questions": 20, "right": 18,
         "tool": "Python"}
for k, v in FACTS.items():
    print(f"{k:10} {v}")
```

```python
WEAK = "Worked on a chatbot using Python."
AI = ("Built an AI chatbot that improved "
      "accuracy by 40%.")
GOOD = ("Built a Python FAQ bot that answered "
        "18 of 20 test questions right, "
        "by retrieving the best chunk.")
for b in (WEAK, AI, GOOD):
    print("-", b)
```

```python
import re

BAD = ("worked on", "responsible for",
       "helped", "involved in")

def checks(b):
    low = b.lower()
    verb = not low.startswith(BAD)
    num = bool(re.search(r"\d", b))
    how = ("using" in low or
           bool(re.search(r" by \w+ing", low)))
    return verb + num + how

for name, b in (("weak", WEAK), ("ai", AI),
                ("good", GOOD)):
    print(name, checks(b), "/ 3")
```

```python
def invented(b):      # number facts mein nahi
    ok = {str(v) for v in FACTS.values()}
    return [n for n in re.findall(r"\d+", b)
            if n not in ok]

for name, b in (("weak", WEAK), ("ai", AI),
                ("good", GOOD)):
    print(name, "invented:", invented(b))
```

### Formula aur AI ka jaal
**X-Y-Z formula** (Laszlo Bock, Google ke purane HR head, apni personal advice mein, 2014): "Accomplished [X] as measured by [Y], by doing [Z]." Yani: kya hasil kiya, kis number se naapa, aur kaise kiya. Ye unki personal advice hai, Google ka hiring rule nahi.

**AI ka jaal:** Anthropic ke docs kehte hain ki model kabhi-kabhi factually galat text likh sakta hai (hallucination). Unke tips: model ko "I don't know" kehne ki ijazat do, quotes se ground karo, verify karo, aur critical baat hamesha khud check karo.

```python
TEST = [True] * 18 + [False] * 2   # 20 sawal
right = sum(TEST)
print(right, "/", len(TEST),
      f"= {right / len(TEST):.0%}")
```

```python
# Apne facts likho, phir apna bullet check karo
MINE = "Developed a tool that cut time by 70%."
print(checks(MINE), "/ 3")
print("invented:", invented(MINE))
```
