Is notebook mein koi AI nahi chalta, koi API key nahi chahiye. Hum ek chhota **checker** banayenge jo batata hai ki email ke prompt mein kya **missing** hai. Neeche wale drafts haath se likhe examples hain.

```python
import re

KISKO = r"\b(manager|client|hr)\b"

def missing(prompt):
    p, m = prompt.lower(), []
    if not re.search(KISKO, p):
        m.append("kisko")
    if not re.search(r"\d", p):     # date/number
        m.append("details")
    if "lines" not in p:
        m.append("kitna")
    return m

print(missing("Chhutti ka email likho"))
```

```python
draft = """Subject: Leave Request
Dear [Manager's Name],
I hope this email finds you well. I am writing
to formally request leave from [Start Date] to
[End Date] due to [Reason]. Before my leave, I
will make sure that all my pending tasks are
completed and that my team is fully informed.
In case of any urgent matter, I will remain
available on email. I sincerely appreciate your
understanding and support in this matter.
Thank you for considering my request.
Warm regards,
[Your Name]"""

khaali = re.findall(r"\[(.*?)\]", draft)
print(len(khaali), "khaali jagah:", khaali)
print(len(draft.split()), "words,",
      len(draft.splitlines()), "lines")
```

```python
TONE = {
    "manager": "seedha aur polite",
    "client":  "polite, par baat saaf",
    "hr":      "formal, dates ke saath",
}
for kisko, tone in TONE.items():
    print(f"{kisko:8} -> {tone}")
```

```python
def email_prompt(kisko, details, lines=5):
    return (f"Mere {kisko} ko email likho. "
            f"Details: {details}. "
            f"Tone: {TONE[kisko]}. "
            f"{lines} lines se kam, subject "
            f"line ke saath.")

p = email_prompt("manager",
    "13-14 Nov chhutti, wajah: behen ki shaadi")
print(p)
print("missing:", missing(p))
```

```python
email = """Subject: 13-14 Nov chhutti
Namaste Sir,
Behen ki shaadi ke liye mujhe 13-14 Nov ki
chhutti chahiye. Mera kaam Rohit dekh lega.
Dhanyavaad, Asha"""

for naam, e in [("draft", draft), ("email", email)]:
    print(f"{naam:5}: {len(e.split()):3} words, "
          f"{len(e.splitlines()):2} lines, "
          f"{e.count('[')} khaali")
```

```python
p2 = email_prompt("client",
    "invoice 214, Rs 45,000, due date 15 Nov")
print(p2)
print("missing:", missing(p2))
```
