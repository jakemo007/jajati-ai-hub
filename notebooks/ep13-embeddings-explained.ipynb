```python
DOCS = [
    "Average package 6 LPA hai",
    "Exam syllabus portal par hai",
    "Canteen mein samosa milta hai",
]
q = "salary kitni milegi"

def keyword(q, doc):   # same words gino
    a = set(q.lower().split())
    return len(a & set(doc.lower().split()))

for d in DOCS:
    print(keyword(q, d), "|", d)
```

```python
# Har word = 3 numbers: [paisa, padhai, khana]
# (haath se rakhe, asli model ke nahi)
VEC = {
    "salary":   [0.9, 0.1, 0.0],
    "package":  [0.7, 0.3, 0.0],
    "ctc":      [0.9, 0.0, 0.1],
    "exam":     [0.0, 0.9, 0.1],
    "syllabus": [0.1, 0.9, 0.0],
    "canteen":  [0.1, 0.1, 0.9],
    "samosa":   [0.0, 0.0, 1.0],
}
print(len(VEC), "words,",
      len(VEC["salary"]), "numbers each")
```

```python
import math

def cos(a, b):       # do vectors ki disha
    dot = sum(x * y for x, y in zip(a, b))
    na = math.sqrt(sum(x * x for x in a))
    nb = math.sqrt(sum(y * y for y in b))
    return dot / (na * nb)

def similar(w1, w2): # 0 se 100
    return round(100 * cos(VEC[w1], VEC[w2]))

for w in ["package", "ctc", "exam", "samosa"]:
    print(f"salary ~ {w:8}: {similar('salary', w)}")
```

```python
def embed(text):     # jo words VEC mein hain
    vs = [VEC[w] for w in text.lower().split()
          if w in VEC]
    return [sum(c) / len(vs) for c in zip(*vs)]

for d in DOCS:
    s = round(100 * cos(embed(q), embed(d)))
    print(f"{s:3} | {d}")
```

### Asli AI ke saath
OpenAI ke docs: "An embedding is a vector (list) of floating point numbers." Asli model ye numbers khud seekhta hai, aur 3 nahi, 1536 ya 3072 deta hai. Similarity naapne ke liye wahi **cosine similarity**.

Kaam: search, clustering, recommendations, classification, aur **RAG** (Episode 4): files ke chunks ko numbers mein badlo, sawal ko bhi, aur sabse paas wale chunks uthao.
