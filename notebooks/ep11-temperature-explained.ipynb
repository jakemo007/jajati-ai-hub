```python
# "Chai ke saath ___": agle word ke score
# (haath se rakhe, asli model ke nahi)
SCORES = {"samosa": 3.0, "biscuit": 2.2,
          "pakoda": 1.6, "pizza": 0.2}
```

```python
import math

def probs(scores, temp=1.0):
    # har score ko temp se divide, phir softmax
    ex = {w: math.exp(s / temp)
          for w, s in scores.items()}
    total = sum(ex.values())
    return {w: e / total for w, e in ex.items()}

def chance(scores, temp=1.0):   # % mein
    return {w: round(100 * p) for w, p
            in probs(scores, temp).items()}

print(chance(SCORES))
```

```python
print(f"{'temp':<4}", *[f"{w:>7}" for w in SCORES])
for t in [0.2, 0.5, 1.0, 2.0]:
    c = chance(SCORES, t).values()
    print(f"{t:<4}", *[f"{v:>6}%" for v in c])
```

```python
import random
from collections import Counter

def bolo(scores, temp):
    if temp == 0:        # 0: hamesha sabse upar
        return max(scores, key=scores.get)
    p = probs(scores, temp)
    words, ws = list(p), list(p.values())
    return random.choices(words, weights=ws)[0]

random.seed(9)          # har baar same result
for t in [0, 0.2, 1.0, 2.0]:
    n = Counter(bolo(SCORES, t)
                for _ in range(20))
    print(f"temp {t}:", ", ".join(
        f"{w} {n[w]}" for w in SCORES if n[w]))
```

```python
# Fact wala sawal: "India ki rajdhani?"
FACT = {"New Delhi": 5.0, "Mumbai": 2.0,
        "Kolkata": 1.6}
for t in [0.2, 1.0, 2.0]:
    print(f"temp {t}:", chance(FACT, t))
```

```python
def do_baar(scores, temp):
    a, b = bolo(scores, temp), bolo(scores, temp)
    print(f"temp {temp}: {a} | {b}")
    if a != b:
        print("  jawab badla: khud check karo!")

random.seed(13)
do_baar(FACT, 0.2)
do_baar(FACT, 2.0)
```
